## 2.2 Kode Program

### 2.2.1 Program "A_2611532027"


```python
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""

print("Hello World")
print()
print("Departemen Informatika")
print("Universitas Andalas")
print("/ \\ // \\\\ /// \\\\\\\\")
print("program ini" + " dibuat oleh a\\n "
      + "azanel adi putro 2611532027.")
```

Tentu, berikut penjelasan **lengkap per baris** sesuai format yang kamu minta:

**Penjelasan :**

1. Bagian pertama kode `"""` berfungsi untuk **memulai komentar** yang dapat terdiri dari beberapa baris.

2. Bagian kedua kode `Komentar` berfungsi sebagai **isi atau keterangan dari komentar** dalam program.

3. Bagian ketiga kode `dua baris atau lebih` berfungsi sebagai **contoh isi komentar yang ditulis dalam beberapa baris**.

4. Bagian keempat kode `pertemuan 1 praktikum alpro` berfungsi untuk **memberikan informasi mengenai kegiatan praktikum**.

5. Bagian kelima kode `"""` berfungsi untuk **mengakhiri komentar**.

6. Bagian keenam kode `print("Hello World")` berfungsi untuk **menampilkan tulisan "Hello World" pada layar**.

7. Bagian ketujuh kode `print()` berfungsi untuk **membuat satu baris kosong pada output program**.

8. Bagian kedelapan kode `print("Departemen Informatika")` berfungsi untuk **menampilkan nama Departemen Informatika**.

9. Bagian kesembilan kode `print("Universitas Andalas")` berfungsi untuk **menampilkan nama Universitas Andalas**.

10. Bagian kesepuluh kode `print("/ \\ // \\\\ /// \\\\\\\\")` berfungsi untuk **menampilkan karakter garis miring (`/`) dan garis miring terbalik (`\`) menggunakan escape character**.

11. Bagian kesebelas kode `print("program ini" + " dibuat oleh a\n " + "azanel adi putro 2611532027.")` berfungsi untuk **menggabungkan beberapa teks menggunakan operator `+` dan membuat baris baru menggunakan `\n`, kemudian menampilkan informasi pembuat program**.
