## 2.1 Dasar Teori

### 2.1.1 If Tunggal
`if` digunakan untuk menjalankan blok perintah ketika suatu kondisi bernilai benar.

### 2.1.2 If-Else
`if-else` menyediakan dua jalur. Jika kondisi pada `if` benar, blok pertama dijalankan; jika salah, blok `else` dijalankan.

### 2.1.3 If-Elif-Else
Struktur ini digunakan untuk beberapa kemungkinan kondisi. Program memeriksa kondisi dari atas ke bawah dan menjalankan blok pertama yang terpenuhi.

### 2.1.4 Operator Logika
Operator `and` mengharuskan semua kondisi yang digabungkan bernilai benar. Operator `!=` digunakan untuk menyatakan tidak sama dengan.

### 2.1.5 Match-Case
`match-case` mencocokkan sebuah nilai dengan beberapa `case`. `case _` digunakan sebagai pilihan default.

### 2.1.6 Multi-If
Beberapa `if` yang terpisah memeriksa kondisi secara independen. Berbeda dengan `elif`, kondisi berikutnya tetap diperiksa walaupun kondisi sebelumnya sudah benar. Pola ini dapat digunakan untuk mengakumulasikan beberapa diskon.


## 2.2 1. Latihan if tunggal — if1_2611532027.py

### Kode Program


```python
ipk_2027 = float(input("Input IPK Anda = "))

if ipk_2027 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2027))

print("Program Selesai")
```

### Analisis Logika
Program menerima IPK dalam bentuk float. Percabangan `if` memeriksa apakah IPK lebih besar dari 2,75. Jika benar, program mencetak keterangan lulus sangat memuaskan. Karena tidak ada `else`, saat kondisi salah program hanya melanjutkan ke `Program Selesai`.

### Analisis Hasil Eksekusi
Input 3,00 membuat kondisi `ipk_2027 > 2.75` bernilai benar sehingga pesan kelulusan ditampilkan.

## 2.3 2. Latihan if-else — if2_2611532027.py

### Kode Program


```python
ipk_2027 = float(input("Input IPK Anda = "))

if ipk_2027 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2027))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

### Analisis Logika
Program masih menggunakan batas IPK 2,75, tetapi sekarang memiliki dua kemungkinan keluaran. Jika kondisi benar, pesan kelulusan ditampilkan; jika salah, bagian `else` dijalankan.

### Analisis Hasil Eksekusi

## 2.4 3. Latihan if-elif-else + operator logika — if_elif_else1_2611532027.py

### Kode Program


```python
umur_2027 = int(input("Input umur anda: "))
sim_2027 = input("Apakah Anda Sudah Punya Sim C: ")[0]

if umur_2027 >= 17 and sim_2027 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_2027 >= 17 and sim_2027!= 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2027 < 17 and sim_2027 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")
print("Program Selesai")
```

### Analisis Logika
Program meminta umur dan status SIM C. Kondisi pertama menggunakan `and`: umur minimal 17 tahun dan SIM bernilai `y`. Kondisi berikutnya membedakan pengunjung dewasa tanpa SIM, pengunjung di bawah 17 tahun yang memasukkan SIM, dan kondisi lainnya.

### Analisis Hasil Eksekusi
Input umur 19 dan SIM `y` memenuhi kondisi pertama, sehingga program mencetak bahwa pengguna sudah dewasa dan boleh membawa motor.

## 2.5 4. Latihan match-case — match_case1_2611532027.py

### Kode Program


```python
bulan_2027 = int(input("Masukkan angka bulan (1 - 12)  "))

match bulan_2027:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

### Analisis Logika
Program mengubah angka bulan menjadi nama bulan. `match` membandingkan nilai `bulan_2027` dengan case 1 sampai 12. `case _` berfungsi sebagai kondisi default ketika angka tidak sesuai dengan 1–12.

### Analisis Hasil Eksekusi
Input 8 memilih `case 8`, sehingga output yang muncul adalah `Agustus`.

## 2.6 5. Latihan multi-if — multi_if1_2611532027.py

### Kode Program


```python
umur_2027 = int(input("Input umur anda: "))
sim_2027 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_2027 >= 17 and sim_2027 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")

if umur_2027 >= 17 and sim_2027!= 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_2027 < 17 and sim_2027 == 'y':
    print("Anda Belum Cukup Umur punya SIM")

if umur_2027 < 17 and sim_2027!= 'y':
    print("Anda Belum Cukup Umur bawa motor")
```

### Analisis Logika
Berbeda dari `if-elif-else`, empat kondisi dibuat sebagai `if` yang berdiri sendiri. Setiap kondisi diperiksa tanpa menghentikan pemeriksaan kondisi berikutnya. Ini menjadi dasar konsep multi-if untuk kondisi yang dapat berdiri secara independen.

### Analisis Hasil Eksekusi
Input umur 19 dan SIM `y` hanya memenuhi kondisi pertama; tiga kondisi lain bernilai salah sehingga hanya satu pesan yang dicetak.

## 2.7 6. Latihan multi-if akumulasi diskon — multi_if2_2611532027.py

### Kode Program


```python
# Input dari user
total_belanja_2027 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2027 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member = input_member_2027 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2027 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2027 = input_promo_2027 in ["y", "ya"]

total_diskon_persen_2027 = 0

if total_belanja_2027 > 1000000:
    total_diskon_persen_2027 += 10  # Diskon belanja besar

if is_member:
    total_diskon_persen_2027 += 5  # Diskon member

if kode_promo_valid_2027:
    total_diskon_persen_2027 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2027 = total_belanja_2027 * (total_diskon_persen_2027 / 100)
total_bayar_2027 = total_belanja_2027 - nominal_diskon_2027

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2027}% (Rp {nominal_diskon_2027:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2027:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2027}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

### Analisis Logika
Program mengubah input member dan promo menjadi nilai boolean. Kemudian empat kondisi diskon diperiksa secara terpisah. Jika syarat terpenuhi, persentase ditambahkan ke `total_diskon_persen_2027`. Karena menggunakan beberapa `if`, diskon dapat ditumpuk.

### Analisis Hasil Eksekusi
Belanja Rp1.500.000 memenuhi diskon 10%, member menambah 5%, dan promo valid menambah 15%. Total diskon menjadi 30%, nominal Rp450.000, dan total bayar Rp1.050.000.

## 2.8 7. Latihan if-elif-else nilai — nilai_2611532027.py

### Kode Program


```python
nilai_2027 = int(input("Inputkan nilai angka= "))

if nilai_2027 >= 81:
    print("A")
elif nilai_2027 >= 70:
    print("B")
elif nilai_2027 >= 60:
    print("C")
elif nilai_2027 >= 50:
    print("D")
else:
    print("E")
```

### Analisis Logika
Program mengelompokkan nilai angka menjadi huruf A sampai E. Pemeriksaan dilakukan dari batas tertinggi ke terendah: ≥81 untuk A, ≥70 untuk B, ≥60 untuk C, ≥50 untuk D, dan selain itu E.

### Analisis Hasil Eksekusi
Input 75 tidak mencapai 81 tetapi mencapai 70, sehingga alur berhenti pada `elif nilai_2027 >= 70` dan menghasilkan nilai B.

## 2.9 8. Tugas 4 — tugas4_2611532027.py

### Kode Program


```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# Input Data Pengunjung
nama_2027 = input("Masukkan Nama Pengunjung        : ")
umur_2027 = int(input("Input umur anda                 : "))
sim_2027 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()
jumlah_tiket_2027 = int(input("Masukkan jumlah tiket           : "))

# If tunggal untuk validasi jumlah tiket
if jumlah_tiket_2027 <= 0:
    print("Peringatan: Kuota tiket tidak valid.")

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_2027 = int(input("Masukkan nomor paket (1-5) : "))

# Match-case untuk memilih wahana
match paket_2027:
    case 1:
        nama_wahana_2027 = "Wahana Safari Rimba"
        harga_satuan_2027 = 50000

    case 2:
        nama_wahana_2027 = "Wahana Arung Jeram"
        harga_satuan_2027 = 75000

    case 3:
        nama_wahana_2027 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2027 = 120000

    case 4:
        nama_wahana_2027 = "Wahana Roller Coaster Kilat"
        harga_satuan_2027 = 100000

    case 5:
        nama_wahana_2027 = "Wahana All-Access VIP"
        harga_satuan_2027 = 220000

    case _:
        print("Paket wahana tidak valid!")
        exit()

# Validasi kelayakan wahana
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2027 == 3 and umur_2027 >= 17 and sim_2027 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")

elif paket_2027 == 3 and umur_2027 >= 17 and sim_2027 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")

elif paket_2027 == 3 and umur_2027 < 17 and sim_2027 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")

elif paket_2027 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")

elif umur_2027 >= 10:
    print("Status Akses: Anda memenuhi syarat umur untuk wahana.")

else:
    print("Status Akses: Anda belum cukup umur untuk wahana ini.")

# Input member dan promo
is_member_2027 = input("\nApakah Anda member? (y/t)      : ").strip().lower()
kode_promo_valid_2027 = input("Apakah kode promo valid? (y/t) : ").strip().lower()

# Menghitung subtotal
subtotal_2027 = harga_satuan_2027 * jumlah_tiket_2027

# Multi-if terpisah untuk akumulasi diskon
total_diskon_persen_2027 = 0

if subtotal_2027 >= 200000:
    total_diskon_persen_2027 += 10

if is_member_2027 in ['y', 'ya']:
    total_diskon_persen_2027 += 5

if kode_promo_valid_2027 in ['y', 'ya']:
    total_diskon_persen_2027 += 15

if jumlah_tiket_2027 >= 5:
    total_diskon_persen_2027 += 5

# Menghitung diskon dan total bayar
nominal_diskon_2027 = subtotal_2027 * (total_diskon_persen_2027 / 100)
total_bayar_2027 = subtotal_2027 - nominal_diskon_2027

# Audit transaksi
print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Nama Pengunjung   : {nama_2027}")
print(f"Wahana            : {nama_wahana_2027}")
print(f"Harga Satuan      : Rp {harga_satuan_2027:,.0f}")
print(f"Jumlah Tiket      : {jumlah_tiket_2027}")
print(f"Subtotal Belanja  : Rp {subtotal_2027:,.0f}")
print(f"Total Diskon      : {total_diskon_persen_2027}% "
      f"(Rp {nominal_diskon_2027:,.0f})")
print(f"Total Bayar       : Rp {total_bayar_2027:,.0f}")

# If-else untuk bonus
if total_bayar_2027 > 300000:
    print("Catatan Layanan   : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan   : Terima kasih telah berkunjung.")

print("\nProgram Selesai")
```

### Analisis Logika
Program menggabungkan materi percabangan menjadi satu sistem loket. `if` tunggal memeriksa jumlah tiket, `match-case` memilih wahana, `if-elif-else` memeriksa kelayakan ATV dengan operator `and` dan `!=`, multi-if menghitung diskon akumulatif, lalu `if-else` menentukan bonus souvenir berdasarkan total pembayaran. Program tidak menggunakan nested-if.

### Analisis Hasil Eksekusi
Pada contoh eksekusi, paket 3 dipilih dengan umur 19 dan SIM `y`, sehingga akses ATV diberikan. Subtotal 2 tiket adalah Rp240.000. Diskon 10% + 5% + 15% = 30%, sehingga total bayar Rp168.000 dan pengguna memperoleh pesan terima kasih.

## 3.1 Alur Logika Program

Alur program tugas dimulai dari input nama, umur, SIM C, dan jumlah tiket. Jumlah tiket divalidasi menggunakan `if` tunggal. Setelah itu pengguna memilih paket wahana melalui `match-case`.

Jika paket yang dipilih adalah Motor ATV Ekstrim, program mengevaluasi umur dan kepemilikan SIM menggunakan `if-elif-else` dengan operator `and` dan `!=`. Untuk paket lain, program menggunakan pengecekan umur sederhana.

Selanjutnya program meminta status member dan kode promo. Subtotal dihitung dari harga satuan dikalikan jumlah tiket. Empat blok `if` terpisah kemudian memeriksa syarat diskon satu per satu sehingga diskon dapat diakumulasikan.

Terakhir, nominal diskon dan total pembayaran dihitung. `if-else` digunakan untuk menentukan apakah total pembayaran lebih dari Rp300.000 sehingga mendapatkan souvenir gratis atau menampilkan ucapan terima kasih.

## 3.2 Analisis Kondisi Diskon

Pada tugas, diskon tidak menggunakan `elif` karena setiap jenis diskon dapat diperoleh secara bersamaan.

| Kondisi | Tambahan Diskon |
|---|---:|
| Subtotal ≥ Rp200.000 | 10% |
| Member `y`/`ya` | 5% |
| Promo valid `y`/`ya` | 15% |
| Jumlah tiket ≥ 5 | 5% |

Pada contoh eksekusi, subtotal Rp240.000 memenuhi diskon belanja besar, status member bernilai `ya`, dan promo bernilai `ya`. Jumlah tiket hanya 2 sehingga diskon rombongan tidak aktif. Total diskon = 10% + 5% + 15% = 30%.

## 3.3 Analisis Hasil Tugas

Dengan data contoh, pengguna memilih Motor ATV Ekstrim sebanyak 2 tiket. Harga satu tiket Rp120.000 sehingga subtotal adalah Rp240.000. Total diskon 30% menghasilkan potongan Rp72.000. Total pembayaran menjadi Rp168.000.
