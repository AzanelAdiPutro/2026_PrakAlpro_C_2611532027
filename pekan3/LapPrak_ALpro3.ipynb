## 2.1 Dasar Teori
Operator dalam Python adalah simbol atau kata yang digunakan untuk melakukan operasi terhadap nilai atau variabel. Operator digunakan dalam berbagai proses, seperti perhitungan, pemberian nilai, perbandingan, dan pengambilan keputusan dalam program. Beberapa jenis operator yang digunakan dalam praktikum ini adalah:

1. Operator Aritmatika
Operator aritmatika digunakan untuk melakukan perhitungan matematika, seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), modulus (%), pangkat (**), dan pembagian bulat (//).
2. Operator Assignment
Operator assignment digunakan untuk memberikan atau mengubah nilai suatu variabel. Contohnya adalah =, +=, -=, *=, dan /=.
3. Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan berupa True atau False. Contohnya adalah ==, !=, >, <, >=, dan <=.
4. Operator Logika
Operator logika digunakan untuk menggabungkan atau membalik suatu kondisi. Operator yang digunakan adalah and, or, dan not.
5. Operator Bitwise
Operator bitwise digunakan untuk melakukan operasi pada tingkat bit atau representasi biner suatu bilangan. Contohnya adalah & (AND), | (OR), ^ (XOR), ~ (NOT), << (left shift), dan >> (right shift).
6. Operator Membership dan Identity
Operator membership digunakan untuk memeriksa apakah suatu nilai terdapat dalam sebuah data, menggunakan in dan not in. Sementara itu, operator identity digunakan untuk memeriksa apakah dua objek merupakan objek yang sama dengan menggunakan is dan is not.

## 2.2 Kode Program

### 2.2.1 Program "C_2611532027"


```python
angka1_2027 = 7
angka2_2027 = 13


hasil = angka1_2027 + angka2_2027
print("\noperator penjumlahan")
print("hasil : ", hasil)

hasil = angka1_2027 - angka2_2027
print("\noperator pengurangan")
print("hasil : ", hasil)

hasil = angka1_2027 * angka2_2027
print("\noperator perkalian")
print("hasil : ", hasil)

if angka2_2027 != 0:
	hasil = angka1_2027 / angka2_2027
	print("\noperator pembagian")
	print("hasil : ", hasil)

	hasil = angka1_2027 // angka2_2027
	print("\noperator pembagian bulat")
	print("hasil : ", hasil)

	hasil = angka1_2027 % angka2_2027
	print("\noperator sisa bagi")
	print("hasil : ", hasil)
else:
	print("angka kedua tidak boleh bernilai 0")

hasil = angka1_2027 ** angka2_2027
print("\noperator pangkat")
print("hasil : ", hasil)

```

**Penjelasan :**

1. Bagian pertama kode ini berfungsi untuk meminta pengguna memasukkan dua bilangan bulat menggunakan `input()`, kemudian mengubah data tersebut menjadi tipe integer dengan `int()` dan menyimpannya ke dalam variabel `angka1_2027` dan `angka2_2027`.

2. Bagian kedua berfungsi untuk melakukan operasi aritmatika dasar, yaitu penjumlahan (`+`), pengurangan (`-`), dan perkalian (`*`). Hasil dari setiap operasi disimpan dalam variabel `hasil` dan ditampilkan menggunakan `print()`.

3. Bagian ketiga berfungsi untuk melakukan operasi pembagian (`/`), pembagian bulat (`//`), dan sisa bagi (`%`). Sebelum melakukan operasi tersebut, program memeriksa apakah `angka2_2027` tidak sama dengan 0 menggunakan operator `!=`. Hal ini dilakukan agar program tidak mengalami pembagian dengan nol.

4. Bagian keempat berfungsi untuk melakukan operasi perpangkatan menggunakan operator `**`. Nilai `angka1_2027` dipangkatkan dengan `angka2_2027`, kemudian hasilnya ditampilkan.

5. Secara keseluruhan, program ini digunakan untuk mempraktikkan berbagai **operator aritmatika Python**, yaitu `+`, `-`, `*`, `/`, `//`, `%`, dan `**`.
