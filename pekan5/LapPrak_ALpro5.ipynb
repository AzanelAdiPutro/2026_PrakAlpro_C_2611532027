# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori

1. **Perulangan `for`**  
   Perulangan `for` digunakan untuk mengulangi blok kode berdasarkan elemen atau nilai yang dihasilkan oleh suatu iterable, salah satunya `range()`.

2. **Fungsi `range()`**  
   Bentuk umum `range(start, stop, step)`. Nilai `stop` tidak termasuk dalam hasil perulangan. Jika `start` tidak ditulis, nilai awalnya adalah 0 dan jika `step` tidak ditulis, nilainya adalah 1.

3. **Accumulator**  
   Accumulator adalah variabel yang digunakan untuk menyimpan hasil perhitungan secara bertahap. Contohnya `jumlah_2027 = jumlah_2027 + i`.

4. **Nested `for`**  
   Nested `for` adalah perulangan `for` yang berada di dalam perulangan `for` lainnya. Perulangan luar biasanya mengatur baris, sedangkan perulangan dalam mengatur isi setiap baris.

5. **Formula matematis**  
   Untuk penjumlahan bilangan 1 sampai `n` digunakan:

   **S = n(n + 1) / 2**

   Jika yang dijumlahkan hanya bilangan genap sampai `n`, untuk `n = 2k` dapat digunakan:

   **S = 2 + 4 + ... + 2k = k(k + 1)**

6. **Kompleksitas nested loop sederhana**  
   Jika dua `for` masing-masing melakukan `n` iterasi, jumlah operasi perulangan bagian dalam secara umum adalah sekitar **n²**.

## 2.2.1 Latihan 1 – Perulangan Dasar

**File program:** `perulangan_for1_2611532027.py`

**Analisis formula matematis:**  
Tidak terdapat perhitungan matematis khusus. Jumlah iterasi ditentukan oleh `ulang_2027`, yaitu sebanyak **n kali** dengan nilai `i = 0, 1, ..., n-1`.

**Alur logika perulangan:**  
Input `n` → `range(n)` dibuat → `i` mengambil nilai satu per satu → perintah `print()` dijalankan → setelah `i = n-1`, perulangan selesai.

```python
ulang_2027 = int(input("Masukkan jumlah perulangan: "))

for i in range(ulang_2027):
    print("Perulangan ke-{i}" )
```

**Penjelasan:**
1. Program menerima jumlah perulangan melalui `ulang_2027`.
2. `range(ulang_2027)` menghasilkan nilai mulai dari 0 sampai `ulang_2027 - 1`.
3. Setiap iterasi menjalankan `print()`.
4. Pada kode yang diberikan, teks `"Perulangan ke-{i}"` tidak menggunakan f-string sehingga `{i}` tampil sebagai teks biasa. Secara alur, variabel `i` tetap berubah pada setiap iterasi.

## 2.2.2 Latihan 2 – Penggunaan `range()`

**File program:** `perulangan_for2_2611532027.py`

**Analisis formula matematis:**  
Untuk `n = 5`, `range(n)` menghasilkan **0 sampai n-1**, sedangkan `range(1, n+1)` menghasilkan **1 sampai n**.

**Alur logika perulangan:**  
Input `n` → perulangan pertama menghasilkan 0..n-1 → cetak hasil → perulangan kedua menghasilkan 1..n → cetak hasil → selesai.

```python
ulang_2027 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_2027-1)
for i in range(ulang_2027):
    print(i, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_2027)
for i in range(1,ulang_2027+1):
    print(i, end=" ")
```

**Penjelasan:**
1. Perulangan pertama menggunakan `range(ulang_2027)`, sehingga dimulai dari 0 dan berhenti sebelum nilai `ulang_2027`.
2. Perulangan kedua menggunakan `range(1, ulang_2027 + 1)`, sehingga dimulai dari 1 dan mencapai `ulang_2027`.
3. Perbedaan `+1` pada batas akhir diperlukan karena nilai `stop` pada `range()` tidak ikut diproses.

## 2.2.3 Latihan 3 – Penjumlahan Bilangan

**File program:** `perulangan_for3_2611532027.py`

**Analisis formula matematis:**  
Secara matematis:

   **S = 1 + 2 + 3 + ... + n**

   atau menggunakan rumus:

   **S = n(n + 1) / 2**

   Untuk `n = 5`, diperoleh `5(6)/2 = 15`.

**Alur logika perulangan:**  
Input `n` → `jumlah = 0` → `i` berjalan 1..n → `jumlah` ditambah `i` pada setiap iterasi → tampilkan operator `+` atau `=` → tampilkan jumlah akhir.

```python
ulang_2027 = int(input("Masukkan jumlah perulangan: "))

jumlah_2027 = 0
for i in range(1, ulang_2027 + 1):
    print(i, end=" ")
    jumlah_2027 = jumlah_2027 + i

    if i < ulang_2027:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2027, end="")
print()
print("Jumlah =", jumlah_2027)
```

**Penjelasan:**
1. `jumlah_2027` diinisialisasi dengan 0.
2. `for` berjalan dari 1 sampai `ulang_2027`.
3. Pada setiap iterasi, nilai `i` ditambahkan ke accumulator: `jumlah_2027 = jumlah_2027 + i`.
4. Kondisi `if` digunakan untuk menampilkan tanda `+` selama belum mencapai bilangan terakhir dan tanda `=` ketika iterasi terakhir.
5. Setelah loop selesai, jumlah akhir dicetak.

## 2.2.4 Latihan 4 – Penjumlahan Bilangan Genap

**File program:** `jumlah_genap_2611532027.py`

**Analisis formula matematis:**  
Jika batas `n` genap dan `n = 2k`, maka:

   **S = 2 + 4 + 6 + ... + 2k**

   **S = 2(1 + 2 + ... + k) = k(k + 1)**

   Untuk `n = 10`, maka `k = 5` dan `S = 5(6) = 30`.

**Alur logika perulangan:**  
Input batas → `i` berjalan 1..n → cek `i % 2 == 0` → jika benar, tambahkan `i` ke jumlah → lanjut ke nilai berikutnya → cetak jumlah akhir.

```python
ulang_2027 = int(input("Masukkan nilai batas: "))

jumlah_2027 = 0
for i in range(1, ulang_2027 + 1):
    if i % 2 == 0:
        print(i, end=" ")
        jumlah_2027 = jumlah_2027 + i

        if i < ulang_2027-1:
            print(" + ", end="")
        else:
            print(" = ", jumlah_2027, end="")
print()
print("Jumlah =", jumlah_2027)
```

**Penjelasan:**
1. Program melakukan iterasi dari 1 sampai nilai batas.
2. Kondisi `i % 2 == 0` memeriksa apakah `i` merupakan bilangan genap.
3. Hanya bilangan genap yang dicetak dan ditambahkan ke `jumlah_2027`.
4. Setelah seluruh nilai diperiksa, jumlah bilangan genap ditampilkan.

## 2.3 Latihan Nested `for`

Selain empat latihan utama, beberapa file latihan yang diberikan menerapkan nested `for`. Bagian ini didokumentasikan karena masih berkaitan langsung dengan konsep perulangan pada pekan ke-5.

### 2.3.1 Nested For 1 – Pola Titik

**File program:** `nested_for1_2611532027.py`

```python
batas_2027 = int(input("Masukkan nilai batas: "))
for line_2027 in range(1, batas_2027 + 1):
    for j_2027 in range(1, (-1 * line_2027 + batas_2027) + 1):
        print(".", end="")
    print(line_2027)
```

**Analisis dan penjelasan:** Perulangan luar menentukan nomor baris. Perulangan dalam mencetak titik sebanyak `batas_2027 - line_2027`. Secara matematis jumlah titik pada baris ke-`i` adalah **n - i**, sehingga jumlah titik semakin berkurang satu setiap baris.

### 2.3.2 Nested For 2 – Pola Persegi

**File program:** `nested_for2_2611532027.py`

```python
batas_2027 = int(input("Masukkan nilai batas: "))
for i_2027 in range(1, batas_2027 + 1):
    for j_2027 in range(1, batas_2027 + 1):
        print("*", end="")
    print()
```

**Analisis dan penjelasan:** Perulangan luar mengatur jumlah baris, sedangkan perulangan dalam mencetak `*` sebanyak `batas_2027` pada setiap baris. Jika batas adalah `n`, maka terdapat **n × n = n²** karakter bintang.

### 2.3.3 Nested For 3 – Penjumlahan Indeks

**File program:** `nested_for3_2611532027.py`

```python
batas_2027 = int(input("Masukkan nilai batas: "))
for i_2027 in range(batas_2027+1):
    for j_2027 in range(batas_2027+1):
        print(i_2027+j_2027, end=" ")
    print()
```

**Analisis dan penjelasan:** Setiap elemen keluaran dihitung menggunakan formula `i + j`. Karena kedua indeks berjalan dari 0 sampai `n`, pola yang terbentuk berupa matriks penjumlahan indeks.

### 2.3.4 Nested For 4 – Pola Khusus

**File program:** `nested_for4_2611532027.py`

```python
tinggi_2027 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2027 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2027 = tinggi_2027
    c_2027 = a_2027
    lebar_2027 = (2 * tinggi_2027) - 2

    for i_2027 in range(1, tinggi_2027 + 1):
        b_2027 = c_2027 + 1
        for j_2027 in range(1, lebar_2027 + 1):
            if i_2027 == 1 or i_2027 == tinggi_2027:
                print("#" if j_2027 == 1 or j_2027 == lebar_2027 else "=", end="")
            else:
                if j_2027 == 1 or j_2027 == lebar_2027:
                    print("|", end="")
                else:
                    if j_2027 == c_2027:
                        print("<", end="")
                    elif j_2027 == b_2027:
                        print(">", end="")
                    elif j_2027 == (lebar_2027 - c_2027):
                        print(">", end="")
                    elif j_2027 == (lebar_2027 - c_2027 + 1):
                        print("<", end="")
                    elif j_2027 > b_2027 and j_2027 < (lebar_2027 - c_2027):
                        print("-", end="")
                    else:
                        print(" ", end="")

        print()
        a_2027 -= 2
        c_2027 = (-a_2027) + 2 if a_2027 <= 0 else a_2027
```

**Analisis dan penjelasan:** Program diawali dengan validasi bahwa tinggi harus genap. Variabel `lebar_2027` dihitung dengan formula **2n - 2**. Nested `for` kemudian mengatur posisi baris dan kolom. Kondisi `if/elif` menentukan karakter yang dicetak berdasarkan posisi kolom. Nilai `a_2027` dan `c_2027` diperbarui setiap baris untuk mengubah posisi pola.

## 2.4 Latihan Tambahan – Segitiga Bintang

**File program:** `latihan5_2611532027.py`

```python
tinggi_2027 = int(input("Masukkan tinggi segitiga: "))

for i_2027 in range(1, tinggi_2027 + 1):
    # 1. cetak spasi
    for s_2027 in range(tinggi_2027 - i_2027):
        print(" ", end="")
    # 2. cetak bintang
    for b_2027 in range(i_2027):
        print("* ", end="")
    print()
```

**Analisis formula dan alur logika:**

Untuk baris ke-`i`, jumlah spasi adalah **n - i**, sedangkan jumlah bintang adalah **i**. Perulangan pertama mencetak spasi agar posisi bintang bergeser ke kiri setiap baris, lalu perulangan kedua mencetak bintang sebanyak nomor baris. Setelah itu pindah ke baris berikutnya.

## Tugas Praktikum

**Soal :**  
Membuat program menggunakan konsep perulangan `for` untuk menghasilkan pola **jam pasir kristal palindromik** berdasarkan ukuran skala `N`.

**Kode Jawaban:**

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_2027 = int(input("Masukkan ukuran skala jam pasir (N): "))

# =========================
# BORDER ATAS
# =========================
print("#", end="")

for i_2027 in range(4 * n_2027 + 5):
    print("=", end="")

print("#")


# =========================
# FASE 1: JAM PASIR ATAS
# =========================
for baris_2027 in range(n_2027, 0, -1):

    print("| ", end="")

    # Spasi penyeimbang kiri
    spasi_2027 = 2 * (n_2027 - baris_2027)

    for i_2027 in range(spasi_2027):
        print(" ", end="")

    # Angka menurun
    for angka_2027 in range(baris_2027, 0, -1):
        print(angka_2027, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Angka menaik
    for angka_2027 in range(1, baris_2027 + 1):
        print(" ", end="")
        print(angka_2027, end="")

    # Spasi penyeimbang kanan
    for i_2027 in range(spasi_2027):
        print(" ", end="")

    print(" |")


# =========================
# FASE 2: POROS PUSAT
# =========================
print("| ", end="")

spasi_2027 = 2 * n_2027 + 1

for i_2027 in range(spasi_2027):
    print(" ", end="")

print("<*>", end="")

for i_2027 in range(spasi_2027):
    print(" ", end="")

print(" |")


# =========================
# FASE 3: JAM PASIR BAWAH
# =========================
for baris_2027 in range(1, n_2027 + 1):

    print("| ", end="")

    # Spasi penyeimbang kiri
    spasi_2027 = 2 * (n_2027 - baris_2027)

    for i_2027 in range(spasi_2027):
        print(" ", end="")

    # Angka menurun
    for angka_2027 in range(baris_2027, 0, -1):
        print(angka_2027, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Angka menaik
    for angka_2027 in range(1, baris_2027 + 1):
        print(" ", end="")
        print(angka_2027, end="")

    # Spasi penyeimbang kanan
    for i_2027 in range(spasi_2027):
        print(" ", end="")

    print(" |")


# =========================
# BORDER BAWAH
# =========================
print("#", end="")

for i_2027 in range(4 * n_2027 + 5):
    print("=", end="")

print("#")
```

**Penjelasan dan analisis formula:**

1. **Border** menggunakan `range(4 * n_2027 + 5)` untuk menentukan panjang garis `=`.
2. **Fase atas** menggunakan `range(n_2027, 0, -1)` sehingga ukuran baris berkurang dari `N` menuju 1.
3. **Spasi kiri dan kanan** menggunakan formula `2 * (n_2027 - baris_2027)` agar bentuk pola tetap seimbang.
4. **Angka menurun** menggunakan `range(baris_2027, 0, -1)`.
5. **Angka menaik** menggunakan `range(1, baris_2027 + 1)`, sehingga bagian kiri dan kanan membentuk pola palindromik.
6. **Fase pusat** hanya menampilkan poros `<*>`.
7. **Fase bawah** mengulangi pola dengan arah ukuran yang berlawanan sehingga terbentuk jam pasir.
8. Alur keseluruhan program adalah: input `N` → border atas → jam pasir atas → poros pusat → jam pasir bawah → border bawah.
